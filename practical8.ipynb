```python
# ============================================================
# PRACTICAL NO. 08
# Smart Surveillance System - Image Classification
# Comparison of ANN and CNN Classifiers using CIFAR-10
# ============================================================

# -----------------------------
# 1. Import Required Libraries
# -----------------------------
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras import models, layers
from sklearn.metrics import confusion_matrix, classification_report

print("TensorFlow Version:", tf.__version__)


# -----------------------------
# 2. Load CIFAR-10 Dataset
# -----------------------------
print("\nLoading CIFAR-10 dataset...")

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Convert labels from (n,1) to (n,)
y_train = y_train.flatten()
y_test = y_test.flatten()

print("Training images:", X_train.shape)
print("Training labels:", y_train.shape)
print("Testing images :", X_test.shape)
print("Testing labels :", y_test.shape)


# -----------------------------
# 3. Define Class Names
# -----------------------------
class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]

print("\nClasses:")
for i, name in enumerate(class_names):
    print(i, "-", name)


# -----------------------------
# 4. Display Sample Images
# -----------------------------
plt.figure(figsize=(12, 8))

for i in range(20):
    plt.subplot(4, 5, i + 1)
    plt.imshow(X_train[i])
    plt.title(class_names[y_train[i]])
    plt.axis('off')

plt.suptitle("Sample Images from CIFAR-10 Dataset", fontsize=16)
plt.tight_layout()
plt.show()


# -----------------------------
# 5. Normalize Image Data
# -----------------------------
X_train_norm = X_train.astype('float32') / 255.0
X_test_norm = X_test.astype('float32') / 255.0

print("\nImage normalization completed.")


# ============================================================
# CLASSIFIER 1: ARTIFICIAL NEURAL NETWORK (ANN)
# ============================================================

print("\n")
print("=" * 60)
print("TRAINING ARTIFICIAL NEURAL NETWORK (ANN)")
print("=" * 60)


# -----------------------------
# 6. Create ANN Model
# -----------------------------
ann_model = models.Sequential([

    layers.Flatten(input_shape=(32, 32, 3)),

    layers.Dense(512, activation='relu'),
    layers.Dense(256, activation='relu'),
    layers.Dense(128, activation='relu'),

    layers.Dense(10, activation='softmax')
])


# -----------------------------
# 7. Compile ANN
# -----------------------------
ann_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nANN Model:")
ann_model.summary()


# -----------------------------
# 8. Train ANN
# -----------------------------
ann_history = ann_model.fit(
    X_train_norm,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)


# -----------------------------
# 9. Evaluate ANN
# -----------------------------
ann_loss, ann_accuracy = ann_model.evaluate(
    X_test_norm,
    y_test,
    verbose=0
)

print("\nANN Test Accuracy: {:.2f}%".format(ann_accuracy * 100))


# ============================================================
# CLASSIFIER 2: CONVOLUTIONAL NEURAL NETWORK (CNN)
# ============================================================

print("\n")
print("=" * 60)
print("TRAINING CONVOLUTIONAL NEURAL NETWORK (CNN)")
print("=" * 60)


# -----------------------------
# 10. Create CNN Model
# -----------------------------
cnn_model = models.Sequential([

    layers.Conv2D(
        32,
        (3, 3),
        activation='relu',
        input_shape=(32, 32, 3)
    ),

    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(
        64,
        (3, 3),
        activation='relu'
    ),

    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(
        128,
        (3, 3),
        activation='relu'
    ),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),

    layers.Dropout(0.5),

    layers.Dense(10, activation='softmax')
])


# -----------------------------
# 11. Compile CNN
# -----------------------------
cnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nCNN Model:")
cnn_model.summary()


# -----------------------------
# 12. Train CNN
# -----------------------------
cnn_history = cnn_model.fit(
    X_train_norm,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)


# -----------------------------
# 13. Evaluate CNN
# -----------------------------
cnn_loss, cnn_accuracy = cnn_model.evaluate(
    X_test_norm,
    y_test,
    verbose=0
)

print("\nCNN Test Accuracy: {:.2f}%".format(cnn_accuracy * 100))


# ============================================================
# 14. ANN vs CNN ACCURACY COMPARISON
# ============================================================

print("\n")
print("=" * 60)
print("ANN vs CNN PERFORMANCE COMPARISON")
print("=" * 60)

print("ANN Test Accuracy : {:.2f}%".format(
    ann_accuracy * 100
))

print("CNN Test Accuracy : {:.2f}%".format(
    cnn_accuracy * 100
))

print("-" * 60)

if cnn_accuracy > ann_accuracy:
    improvement = ((cnn_accuracy - ann_accuracy) / ann_accuracy) * 100

    print("CNN performed better than ANN.")
    print("CNN improvement: {:.2f}%".format(improvement))

else:
    print("ANN performed better than CNN.")


# ============================================================
# 15. PLOT TRAINING AND VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(14, 5))


# ANN Accuracy
plt.subplot(1, 2, 1)

plt.plot(
    ann_history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    ann_history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title('ANN Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)


# CNN Accuracy
plt.subplot(1, 2, 2)

plt.plot(
    cnn_history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    cnn_history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title('CNN Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()


# ============================================================
# 16. CNN PREDICTIONS
# ============================================================

print("\nGenerating CNN predictions...")

cnn_predictions = cnn_model.predict(
    X_test_norm,
    verbose=0
)

# Get class with highest probability
cnn_pred_classes = np.argmax(
    cnn_predictions,
    axis=1
)


# ============================================================
# 17. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    cnn_pred_classes
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')
plt.title('CNN Confusion Matrix')

plt.show()


# ============================================================
# 18. CLASSIFICATION REPORT
# ============================================================

print("\n")
print("=" * 60)
print("CNN CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        cnn_pred_classes,
        target_names=class_names
    )
)


# ============================================================
# 19. RECOGNITION CAPABILITY
# ============================================================

correct = np.sum(
    cnn_pred_classes == y_test
)

total = len(y_test)

incorrect = total - correct

recognition_rate = (
    correct / total
) * 100

print("\n")
print("=" * 60)
print("CNN RECOGNITION CAPABILITY")
print("=" * 60)

print("Total test images       :", total)
print("Correctly recognized    :", correct)
print("Incorrectly recognized  :", incorrect)
print("Recognition Capability  : {:.2f}%".format(
    recognition_rate
))


# ============================================================
# 20. DISPLAY ACTUAL vs PREDICTED IMAGES
# ============================================================

plt.figure(figsize=(14, 10))

for i in range(20):

    plt.subplot(4, 5, i + 1)

    plt.imshow(X_test[i])

    actual = class_names[y_test[i]]

    predicted = class_names[
        cnn_pred_classes[i]
    ]

    # Green = Correct
    # Red = Incorrect
    if actual == predicted:
        color = 'green'
    else:
        color = 'red'

    plt.title(
        "Actual: " + actual +
        "\nPredicted: " + predicted,
        color=color,
        fontsize=9
    )

    plt.axis('off')

plt.suptitle(
    "CNN Object Recognition Results",
    fontsize=16
)

plt.tight_layout()
plt.show()


# ============================================================
# 21. CLASS-WISE ACCURACY
# ============================================================

print("\n")
print("=" * 60)
print("CLASS-WISE RECOGNITION ACCURACY")
print("=" * 60)

for i, class_name in enumerate(class_names):

    class_indices = np.where(y_test == i)[0]

    class_correct = np.sum(
        cnn_pred_classes[class_indices] == i
    )

    class_accuracy = (
        class_correct / len(class_indices)
    ) * 100

    print(
        "{:<12} : {:.2f}%".format(
            class_name,
            class_accuracy
        )
    )


# ============================================================
# 22. FINAL RESULT
# ============================================================

print("\n")
print("=" * 60)
print("FINAL RESULT")
print("=" * 60)

print(
    "ANN Accuracy = {:.2f}%".format(
        ann_accuracy * 100
    )
)

print(
    "CNN Accuracy = {:.2f}%".format(
        cnn_accuracy * 100
    )
)

print(
    "CNN Recognition Capability = {:.2f}%".format(
        recognition_rate
    )
)

if cnn_accuracy > ann_accuracy:
    print("\nConclusion: CNN performed better than ANN")
    print("for image classification and object recognition.")

else:
    print("\nConclusion: ANN performed better than CNN.")

print("\nPractical No. 08 Completed Successfully!")
```
